```python
# convert Order Date and Ship Data to to_datetime because they are object(text)
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
# change the header name to lowecase
df.columns = df.columns.str.lower()
df.columns
```

```python
# replace space with underscore
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'sub-category': 'sub_category'})
df.columns
```

```python
# outliers (weird values)
# look for negative quantity and sales, discount should be in rang 0-100%,, extremely unusual profit values (very negative)
df[['sales', 'quantity', 'discount', 'profit']].describe()
```

```python
# extract the big loss profit
df.loc[df['profit'].idxmin()]
```

```python
# extract the biggest sale
df.loc[df['sales'].idxmax()]
```

```python
# how many orders actually lose money
(df['profit'] < 0).sum()    # return true false and then add up true
```

```python
# does giving customers larger discount affect profitability?
# create discount range
discount_range = pd.cut(df['discount'], bins=[0, 0.2, 0.4, 0.6, 0.8])
# group by those range and calculate the profit for each discount range
df.groupby(discount_range)['profit'].mean()
```
